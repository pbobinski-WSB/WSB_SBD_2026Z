### 1. Przygotowanie "żywych" danych w systemie CRM (Postgres)
Zasymulujemy działanie aplikacji. Wrzucimy do naszej małej, relacyjnej bazy 3 klientów VIP.

```python
import pandas as pd
from sqlalchemy import create_engine

# Łączymy się z systemowym Postgresem
pg_engine = create_engine('postgresql://admin:sekret@postgres_db:5432/firmadb')

# Tworzymy 3 klientów VIP w Pythonie
crm_data = pd.DataFrame({
    'custkey': [1, 2, 3], # ID klienta musi pasować do naszych dużych danych
    'vip_status': ['GOLD', 'PLATINUM', 'DIAMOND'],
    'account_manager': ['Jan Kowalski', 'Anna Nowak', 'Piotr Wiśniewski']
})

# Zapisujemy ich do Postgresa do tabeli 'crm_users'
crm_data.to_sql('crm_users', pg_engine, if_exists='replace', index=False)

# Sprawdzamy, czy się udało
print("Dane w bazie relacyjnej (Postgres):")
print(pd.read_sql("SELECT * FROM crm_users", pg_engine))
```
